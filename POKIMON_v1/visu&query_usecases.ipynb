# Notebook to query and visualize POKIMON instances

```python
# import the vis package
from pckg.ontovis import *
import os 
```

```python
# define Colors for types of instances
color_map = {
    # ICEs
"POK:Semantic_Content_Entity": "#5D7A8C",      
"POK:Depiction_Content_Entity": "#188dd6",   
"POK:Proposition_Content_Entity": "#188dd6",      
"POK:Figurative_Content_Entity": "#188dd6",      

"POK:Conceptual_Geological_Model": "#8bcc55",      
"POK:Space_Time_Geological_Model": "#6fcc55",     

"POK:Action_Specification": "#1780D6", # red


    # Geoentities
"POK:Geological_Unit" :  "#ccb655",  
"POK:Lithology":  "#ccb655",  
"POK:Spatial_Location":  "#ccb655",  
"POK:Temporal_Location":  "#ccb655",  


"Unknown": "#9E9E9F"
}

output_folder = r'.\html_sparql_outputs'  # update if needed
```

```python
# remidner for querries structure as the network graph module is senstive to the sparql structure (e.g., it does not support rdfs:subclasses)
# for exhaustive querying, use a dedicated sparql tool

POKI = Namespace("http://www.POKIMON#")

general_querry_structure = """

PREFIX poki: <http://www.POKIMON#>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT DISTINCT ?subject ?predicate ?object
WHERE {
  ?subject rdf:type poki:Algorithm .

  {
    ?subject poki:is_About ?object1 .
    ?object1 rdf:type poki:Planned_Process .
  }
  FILTER EXISTS {
    ?subject poki:has_Part ?object2 .
    ?object2 rdf:type poki:Action_Specification .
  }

  {
    ?subject poki:is_About ?object .
    BIND(poki:is_About AS ?predicate)
  }
  UNION
  {
    ?subject poki:has_Part ?object .
    BIND(poki:has_Part AS ?predicate)
  }
}

"""
```

## Use case 1 

```python
# import the use_case_1 
## ensure to have the ontology  as an rdf/xml file
pokimon_vis = OntoVis( ontologies_pathes= [r'.\examples\pokimon_use_case_1.owl', ], color_map_instance = color_map)
```

```python
# use case 1 querry all instances

query_1_all_instances = """
PREFIX poki: <http://www.POKIMON#POK:>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
SELECT DISTINCT ?subject ?predicate ?object
WHERE {
  ?subject rdf:type ?class .
  #?object rdf:type ?class2 .
  ?subject ?predicate ?object .
}
"""

# use case 1 querry instances of geomodels (conceptual and space time)


query_1_geomodels= """
PREFIX poki: <http://www.POKIMON#POK:>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT DISTINCT ?subject ?predicate ?object
WHERE {
  VALUES ?targetClass { poki:Conceptual_Geological_Model poki:Space_Time_Geological_Model }
  ?subject rdf:type ?targetClass .

  {
    ?subject poki:has_Part ?object .
    BIND(poki:has_Part AS ?predicate)
  }
  UNION
  {
    ?subject poki:satisfies ?object .
    BIND(poki:satisfies AS ?predicate)
  }
}
  """


# use case 1 querry instances of semantic components of the semantic content entities of geomodels


query_1_smnt_compo = """
PREFIX poki: <http://www.POKIMON#POK:>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT DISTINCT ?subject ?predicate ?object
WHERE {
  ?subject rdf:type ?Semantic_Content_Entity .
  {
    ?subject poki:has_Semantic_Component ?object .
    BIND(poki:has_Semantic_Component AS ?predicate)
  }
}
  """




```

```python
#generate the graphes as html files 
# ensure the directory for outputs is created

pokimon_vis.generate_html_from_network_graph(query_1_all_instances, output_html_path= os.path.join(output_folder, 'UC1_all_instances.html'),  edge_color = 'black', size_node = 60, edge_font_size = 12  )
pokimon_vis.generate_html_from_network_graph(query_1_geomodels, output_html_path= os.path.join(output_folder, 'UC1_geomodels.html'), edge_color = 'black', size_node = 60,  edge_font_size = 12 )
pokimon_vis.generate_html_from_network_graph(query_1_smnt_compo,output_html_path= os.path.join(output_folder, 'UC1_semnt_compo.html'),  edge_color = 'black', size_node = 60,  edge_font_size = 12 )
```
